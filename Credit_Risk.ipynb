```python
#Build PD model
#Descriptive analysis to predictive
#Logistic regression
features = [
    "LIMIT_BAL",
    "AGE",
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6",
    "BILL_AMT1",
    "PAY_AMT1",
    "num_late_payments"
]

X = df[features]
y = df["default"]
```

```python
X.head()
```

```python
print(X.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training observations:", len(X_train))
print("Testing observations:", len(X_test))

print("\nTraining default rate:")
print(y_train.mean())

print("\nTesting default rate:")
print(y_test.mean())
```

```python
#Standarize variables
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
scaler.fit_transform(X_train)
scaler.transform(X_test)
```

```python
#PD model
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=1000
)

model.fit(X_train_scaled, y_train)
```

```python
#Generate predictions
y_pred = model.predict(X_test_scaled)
```

```python
y_prob = model.predict_proba(X_test_scaled)[:, 1]
```

```python
#Evaluate the model
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
auc = roc_auc_score(y_test, y_prob)

print(f"Accuracy:  {accuracy:.3f}")
print(f"Precision: {precision:.3f}")
print(f"Recall:    {recall:.3f}")
print(f"F1 Score:  {f1:.3f}")
print(f"ROC-AUC:   {auc:.3f}")
```

```python
#Threshold analysis 
results = []

for threshold in np.arange(0.10, 0.71, 0.05):

    predictions = (y_prob >= threshold).astype(int)

    results.append({
        "Threshold": threshold,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(y_test, predictions),
        "Recall": recall_score(y_test, predictions),
        "F1": f1_score(y_test, predictions)
    })

threshold_results = pd.DataFrame(results)

threshold_results.round(3)
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.metrics import roc_auc_score

def evaluate_logistic(feature_list):

    X_model = df[feature_list]
    y_model = df["default"]

    X_train_m, X_test_m, y_train_m, y_test_m = train_test_split(
        X_model,
        y_model,
        test_size=0.20,
        random_state=42,
        stratify=y_model
    )

    pipeline = make_pipeline(
        StandardScaler(),
        LogisticRegression(max_iter=1000)
    )

    pipeline.fit(X_train_m, y_train_m)

    probabilities = pipeline.predict_proba(X_test_m)[:, 1]

    auc_score = roc_auc_score(
        y_test_m,
        probabilities
    )

    gini_score = 2 * auc_score - 1

    return auc_score, gini_score
```

```python
#Model 1 Raw repayment history
raw_features = [
    "LIMIT_BAL",
    "AGE",
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6",
    "BILL_AMT1",
    "PAY_AMT1"
]

auc_raw, gini_raw = evaluate_logistic(raw_features)

print(f"AUC:  {auc_raw:.3f}")
print(f"Gini: {gini_raw:.3f}")
```

```python
#Model 3 Combined
combined_features = [
    "LIMIT_BAL",
    "AGE",
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6",
    "BILL_AMT1",
    "PAY_AMT1",
    "num_late_payments"
]

auc_combined, gini_combined = evaluate_logistic(combined_features)

print(f"AUC:  {auc_combined:.3f}")
print(f"Gini: {gini_combined:.3f}")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.metrics import roc_auc_score
import pandas as pd


def evaluate_logistic(feature_list):

    X_model = df[feature_list]
    y_model = df["default"]

    X_train_m, X_test_m, y_train_m, y_test_m = train_test_split(
        X_model,
        y_model,
        test_size=0.20,
        random_state=42,
        stratify=y_model
    )

    pipeline = make_pipeline(
        StandardScaler(),
        LogisticRegression(max_iter=1000)
    )

    pipeline.fit(X_train_m, y_train_m)

    probabilities = pipeline.predict_proba(X_test_m)[:, 1]

    auc = roc_auc_score(y_test_m, probabilities)
    gini = 2 * auc - 1

    return auc, gini


# MODEL 1: Raw repayment variables
raw_features = [
    "LIMIT_BAL",
    "AGE",
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6",
    "BILL_AMT1",
    "PAY_AMT1"
]


# MODEL 2: Engineered repayment variable
engineered_features = [
    "LIMIT_BAL",
    "AGE",
    "BILL_AMT1",
    "PAY_AMT1",
    "num_late_payments"
]


# MODEL 3: Combined
combined_features = [
    "LIMIT_BAL",
    "AGE",
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6",
    "BILL_AMT1",
    "PAY_AMT1",
    "num_late_payments"
]


# Evaluate all three models
auc_raw, gini_raw = evaluate_logistic(raw_features)

auc_eng, gini_eng = evaluate_logistic(engineered_features)

auc_combined, gini_combined = evaluate_logistic(combined_features)


# Create comparison table
model_comparison = pd.DataFrame({
    "Model": [
        "Raw Repayment Features",
        "Engineered Feature",
        "Combined"
    ],
    "AUC": [
        auc_raw,
        auc_eng,
        auc_combined
    ],
    "Gini": [
        gini_raw,
        gini_eng,
        gini_combined
    ]
})

model_comparison.round(3)
```

```python
#Random Forest
from sklearn.ensemble import RandomForestClassifier

X_rf = df[combined_features]
y_rf = df["default"]

X_train_rf, X_test_rf, y_train_rf, y_test_rf = train_test_split(
    X_rf,
    y_rf,
    test_size=0.20,
    random_state=42,
    stratify=y_rf
)

rf_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=8,
    min_samples_leaf=20,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_rf, y_train_rf)
```

```python
rf_prob = rf_model.predict_proba(X_test_rf)[:, 1]
```

```python
rf_pred = rf_model.predict(X_test_rf)
```

```python
rf_auc = roc_auc_score(y_test_rf, rf_prob)
rf_gini = 2 * rf_auc - 1

rf_accuracy = accuracy_score(y_test_rf, rf_pred)
rf_precision = precision_score(y_test_rf, rf_pred)
rf_recall = recall_score(y_test_rf, rf_pred)
rf_f1 = f1_score(y_test_rf, rf_pred)

print(f"Accuracy:  {rf_accuracy:.3f}")
print(f"Precision: {rf_precision:.3f}")
print(f"Recall:    {rf_recall:.3f}")
print(f"F1 Score:  {rf_f1:.3f}")
print(f"ROC-AUC:   {rf_auc:.3f}")
print(f"Gini:      {rf_gini:.3f}")
```

```python
final_comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest"
    ],
    "ROC-AUC": [
        auc_combined,
        rf_auc
    ],
    "Gini": [
        gini_combined,
        rf_gini
    ]
})

final_comparison.round(3)
```
